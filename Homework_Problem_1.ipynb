```python
# sWAP cASE

def swap_case(s):
    result = ""

    for char in s:
        if char.islower():
            result += char.upper()
        elif char.isupper():
            result += char.lower()
        else:
            result += char

    return result


if __name__ == '__main__':
    s = input()
    print(swap_case(s))
```

```python
# Find a String

def count_substring(string, sub_string):
    count = 0

    for i in range(len(string) - len(sub_string) + 1):
        if string[i:i + len(sub_string)] == sub_string:
            count += 1

    return count

if __name__ == '__main__':
    string = input().strip()
    sub_string = input().strip()
    
    count = count_substring(string, sub_string)
    print(count)
```

```python
# Alphabet Rangoli

def print_rangoli(size):
    alphabet = "abcdefghijklmnopqrstuvwxyz"
    letters = []

    for i in range(size):
        letters.append(alphabet[i])

    row_levels = list(range(size))
    row_levels += row_levels[-2::-1]

    width = 4 * size - 3

    for level in row_levels:
        section = letters[-(level + 1):]
        pattern = section[::-1] + section[1:]
        line = "-".join(pattern)

        print(line.center(width, "-"))

if __name__ == '__main__':
    n = int(input())
    print_rangoli(n)
```

```python
# Time Delta

#!/bin/python3
from datetime import datetime
import math
import os
import random
import re
import sys

# Complete the time_delta function below.
def time_delta(t1, t2):
    
    time1 = datetime.strptime(t1, "%a %d %b %Y %H:%M:%S %z")
    time2 = datetime.strptime(t2, "%a %d %b %Y %H:%M:%S %z")

    difference = abs((time1 - time2).total_seconds())

    return str(int(difference))
    

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    t = int(input())

    for t_itr in range(t):
        t1 = input()

        t2 = input()

        delta = time_delta(t1, t2)

        fptr.write(delta + '\n')

    fptr.close()
```

```python
#Closures and decorators

def wrapper(f):
    def fun(l):
        formatted_list = []
        for n in l:
            noprefix = n[-10:]
            first_half = noprefix[:5]
            sec_half = noprefix[5:]
            formatted_n = "+91 " + first_half + " " + sec_half
            formatted_list.append(formatted_n)
        f(formatted_list)
    return fun

@wrapper
def sort_phone(l):
    print(*sorted(l), sep='\n')

if __name__ == '__main__':
    l = [input() for _ in range(int(input()))]
    sort_phone(l) 
```

```python
# Decorators 2 - Name Directory

import operator

def person_lister(f):
    def inner(people):
        people.sort(key=lambda person: int(person[2]))

        result = []

        for person in people:
            result.append(f(person))

        return result

    return inner

@person_lister
def name_format(person):
    return ("Mr. " if person[3] == "M" else "Ms. ") + person[0] + " " + person[1]

if __name__ == '__main__':
    people = [input().split() for i in range(int(input()))]
    print(*name_format(people), sep='\n')
```

```python
# Mean Var and Std

import numpy

size = list(map(int, input().split()))

lst = []

for n in range(size[0]):
    rows = list(map(int, input().split()))
    lst.append(rows)

arr = numpy.array(lst)

print(numpy.mean(arr, axis=1))
print(numpy.var(arr, axis=0))
print(round(numpy.std(arr),11))
```

```python
# Dot and cross

import numpy

size = int(input())

lstA = []
lstB = []

for n in range(size):
    row = list(map(int,input().split()))
    lstA.append(row)
    
for n in range(size):
    row = list(map(int,input().split()))
    lstB.append(row)

arrA = numpy.array(lstA)
arrB = numpy.array(lstB)

print(numpy.dot(arrA, arrB))
```

```python
# Inner and Outer

import numpy

A = list(map(int, input().split()))
B = list(map(int, input().split()))

print(numpy.inner(A,B))
print(numpy.outer(A,B))
```

```python
# Polynomials

import numpy

P = list(map(float,input().split()))
x = float(input())

arrP = numpy.array(P)

print (numpy.polyval(arrP, x))
```

```python
# Linear Algebra

import numpy
dim = int(input())
matrix = []

for n in range(dim):
    
    row = list(map(float,input().split()))
    matrix.append(row)
    
arr = numpy.array(matrix)

print(round(numpy.linalg.det(arr), 2))
```
